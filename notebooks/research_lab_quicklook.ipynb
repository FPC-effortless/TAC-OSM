# TAC-OSM Research Lab — read-only run quicklook
This notebook reads locally stored run receipts only. It does not train, query private TEST data, or establish scientific qualification.

```python
from pathlib import Path
import json
ROOT = Path.cwd()
if not (ROOT / 'runs').exists() and (ROOT.parent / 'runs').exists():
    ROOT = ROOT.parent
runs = sorted((ROOT / 'runs' / 'lab').glob('*/receipt.json'))
print('Recorded runs:', len(runs))
for path in runs[:30]:
    data = json.loads(path.read_text())
    assert data['capability_claim_authorized'] is False
    print(data['run_id'], data['experiment'], data['variant'], data['metrics'])
```
